## Importa las librerías pertinentes para solucionar el reto.

```python
import pandas as pd
import numpy as np
```

## Extrae la información del archivo.

```python
df = pd.read_excel("hospitalData.xlsx")
df.head()
```

## Muestra la estructura y los tipos de datos de cada columna para identificar qué operaciones puedes realizar con cada una de ellas, asegúrandote de que las columnas con fechas sean del tipo datetime64.

```python
df["Week Ending Date"] = pd.to_datetime(df["Week Ending Date"])
df.info()
```

## Elimina los registros con el código geográfico MP.

```python
df = df[df["Geographic aggregation"] != "MP"]
```

# Para las columnas que tienen que ver con la ICU, cambia los valores faltantes por 0.

```python
df[["Number of ICU Beds","Number of Adult ICU Beds","Number of Pediatric ICU Beds","Number of ICU Beds Occupied","Number of Adult ICU Beds Occupied","Number of Pediatric ICU Beds Occupied"]] = df[["Number of ICU Beds","Number of Adult ICU Beds","Number of Pediatric ICU Beds","Number of ICU Beds Occupied","Number of Adult ICU Beds Occupied","Number of Pediatric ICU Beds Occupied"]].fillna(0)
df.info()
```

# Conoce el periodo de registro que se encuentra en el archivo para saber cuántos días han transcurrido entre la fecha más reciente y la más antigua.

```python
df["Week Ending Date"].max() - df["Week Ending Date"].min()
```

# Obtén la cantidad total de pacientes pediátricos hospitalizados con COVID-19.

```python
ped_covid = df["Number of Pediatric Patients Hospitalized with COVID-19"].sum()
print(f"Total de paciented pediatricos hospitalizados con COVID-19: {ped_covid}")
```

# Obtén el promedio de pacientes adultos hospitalizados por RSV.

```python
adult_mean = df["Number of Adult Patients Hospitalized with COVID-19"].mean()
print(f"Promedio de pacientes adultos hospitalizados con COVID-19: {adult_mean.round(2)}")
```

# Agrupa por región geográfica y obtén la sumatoria de pacientes adultos hospitalizados con influenza.

```python
df['Number of Adult Patients Hospitalized with Influenza'].groupby(df['Geographic aggregation']).sum()
```

# Crea una columna derivada llamada "Percent of Adult ICU Beds Occupied" con el porcentaje de ocupación para las camas de los adultos en la ICU.

```python
df['Percent of Adults ICU Beds Occupied'] = df['Number of Adult ICU Beds Occupied'] / df['Number of Adult ICU Beds'] * 100
df['Percent of Adults ICU Beds Occupied']
```

# Crea un *dataframe* llamado "filtro", con las filas de los hospitales que tengan más de 300 camas "disponibles" para adultos en la ICU; y las columnas de identificación geográfica y las relacionadas, con la cantidad de pacientes con alguna de las enfermedades respiratorias de los adultos.

```python
filtro = df.loc[
    df['Number of Adult ICU Beds'] - df['Number of Adult ICU Beds Occupied'] > 300,
    [
        'Geographic aggregation',
        'Number of Adult ICU Beds',
        'Number of Adult ICU Beds Occupied',
        'Number of Adult Patients Hospitalized with COVID-19',
        'Number of Adult Patients Hospitalized with Influenza'
    ]
]
filtro
```

# Crea un *dataframe* llamado "registros2024" que tenga todos los registros del año 2024.

```python
registros2024 = df[df['Week Ending Date'].dt.year == 2024]
registros2024
```

# Crear una columna derivada con el conteo de la cantidad de certificaciones que tiene cada hospital; es decir, cuenta la cantidad de elementos que hay en la columna de certificaciones separándolas con /

```python
df['Cert_Counter'] = df['Certifications'].str.split('/').str.len()
df
```

# Busca todos los hospitales que cumplan con la certificación CRN y déjalos en un *dataframe* llamado "CRNData".

```python
CRNData = df[df['Certifications'].str.contains('CRN')]
CRNData
```

# Almacena el *dataframe* modificado y los *dataframes* creados en los pasos anteriores en un mismo archivo de Excel llamado "datosHospitalesProcesados.xlsx". Cada *dataframe* deberá estar en una hoja diferente.

```python

with pd.ExcelWriter('datosHospitalesProcesados.xlsx', engine='xlsxwriter') as writer:
    df.to_excel(writer, sheet_name='Original', index=False)
    filtro.to_excel(writer, sheet_name='Filtro', index=False)
    registros2024.to_excel(writer, sheet_name='Registro 2024', index=False)
    CRNData.to_excel(writer, sheet_name='CRNData', index=False)
```
