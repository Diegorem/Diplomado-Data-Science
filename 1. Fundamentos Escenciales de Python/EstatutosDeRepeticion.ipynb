# Estatutos de Repeticion


## While

```python
# Estatuto while simple
num = int(input("Dame un entero positivo: "))
i = 0
while i < num:
  print(i)
  i = i + 1
print("Primera línea fuera del while")
```

### Contador

```python
# Uso de un contador
contador_pares = 0
numero = int(input("Ingresa un número positivo (o un negativo para salir): "))
while numero >= 0:
    if numero % 2 == 0:
       contador_pares += 1  # Incrementar el contador si el número es par3
    numero = int(input("Ingresa otro número: "))    
    print("Cantidad de números pares ingresados:", contador_pares)
```

### Acumulador

```python
# Uso de un acumulador para calcular un promedio
suma_calificaciones = 0  # Acumulador
cantidad = 0  # Contador
calificacion = float(input('Ingresa una calificación (o un número negativo para terminar): '))
while calificacion >= 0:
   suma_calificaciones = suma_calificaciones + calificacion  # Acumular la suma de calificaciones
   cantidad = cantidad + 1  # Contar el número de calificaciones ingresadas
   calificacion = float(input('Ingresa otra calificación: '))
if cantidad > 0:
   promedio = suma_calificaciones / cantidad
   print('El promedio de calificaciones es:', promedio)
else:
   print('No se ingresaron calificaciones válidas.')
```

### Bandera

```python
# Uso de una bandera para detener un ciclo while
encontrado_mayor_100 = False  # Bandera inicializada en False
numero = int(input('Ingresa un número positivo (o un negativo para salir): '))
while numero >= 0 and not encontrado_mayor_100:
   if numero > 100:
       encontrado_mayor_100 = True  # Cambiamos la bandera a True para detener el ciclo
   else:
       numero = int(input('Ingresa otro número: '))
if encontrado_mayor_100:
   print('Se ingresó un número mayor a 100, ciclo detenido.')
else:
   print('No se ingresaron números mayores a 100.')
```
