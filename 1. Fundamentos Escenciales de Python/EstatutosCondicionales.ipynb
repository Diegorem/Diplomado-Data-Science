# Estatutos condicionales

### Condicional if

```python
# Ejemplo de condicional if 
temperature = 50 
if temperature >= 38: 
    print('Tienes fiebre')
print('Primera línea después del if')
```

### If-Else

```python
total_compra = 1200  
if total_compra > 1000: 
    print("Se aplica un descuento del 10%.")
else: 
    print("No se aplica ningún descuento.")
print("Gracias por tu compra.")
```

### If-Elif-Else

```python
# Ejemplo del estatuto condicional if - elif - else 
total_compra = 3200 
if total_compra > 3000: 
    print("Se aplica un descuento del 20%.")
elif total_compra > 1000: 
    print("Se aplica un descuento del 10%.")
else: 
    print("No se aplica ningún descuento.")
print("Gracias por tu compra.")
```
