# Tipos y Estructuras de Datos

```python
nombre = "dieGo "
print(nombre.capitalize()) #Capitaliza la primera letra
print(nombre.count("o")) #Cuenta cuantas veces aparece la letra "o"
print(nombre.find("G")) #Busca la letra "G" y devuelve su posición, si no la encuentra devuelve -1
print(nombre.lower()) #Convierte todas las letras a minúsculas
print(nombre.upper()) #Convierte todas las letras a mayúsculas
print(nombre.replace("o", "u")) #Reemplaza la letra "o" por la letra "u"
split = nombre.split("e") #Divide el string en una lista utilizando la letra "e" como separador
print(split)
print(nombre.startswith("d")) #Devuelve True si el string empieza con la letra "d"
print(nombre.strip()) #Elimina los espacios en blanco al principio y al final del string
print(nombre.swapcase()) #Conviete las letras mayúsculas en minúsculas y viceversa
print(nombre.title()) #Convierte la primera letra de cada palabra en mayúscula
```

```python
texto = 'Python es un lenguaje de programación creado por Guido van Rossum. ' 
print(texto.capitalize()) 
print(texto.count('es')) 
print(texto.find('Python')) 
print(texto.lower()) 
print(texto.replace('creado', 'desarrollado')) 
print(texto.split()) 
print(texto.startswith('Python')) 
print(texto.strip()) 
print(texto.swapcase()) 
print(texto.title()) 
print(texto.upper())
```

# Estructuras de datos

Lista

```python
lst = ['azul', 'rojo', 'blanco', 'amarillo', 'azul'] 
print(lst) 

# Obtener la longitud 
print('La lista contiene', len(lst), 'elementos')  

# Recuperar un elemento por su índice 
print(lst[1]) 
print(lst[-2]) 

# Slicing 
print(lst[1:3]) 
print(lst[:3]) 
print(lst[1:]) 
print(lst[1:4:2]) 

# Agregar elementos 
lst.append('morado') 
lst.insert(2,'rosa') 
print(lst) 

# Quitar elementos 
lst.remove('blanco')
lst.pop()
print(lst)
```

Tupla

```python
# Tupla 
tup = ('azul', 'rojo', 'blanco', 'amarillo', 'azul') 
print(tup) 

# Obtener la longitud 
print('La tupla contiene', len(tup), 'elementos') 

# Recuperar un elemento por su índice 
print(tup[1])
print(tup[-2]) 

# Slicing 
print(tup[1:3]) 
print(tup[:3]) 
print(tup[1:]) 
print(tup[1:4:2]) 
```

Diccionario

```python
# Diccionario 
mi_perro = {'raza': 'pastor alemán', 'nombre': 'Linus', 'peso': 45} 
print(mi_perro) 

# Obtener la longitud 
print('El diccionario contiene', len(mi_perro), 'elementos') 

# Recuperar un valor por su clave 
print(mi_perro['raza']) 
print(mi_perro.get('nombre')) 

# Modificar un valor 
mi_perro['peso'] = 40 
print(mi_perro) 

# Agregar un valor 
mi_perro['color'] = 'negro' 
print(mi_perro) 

# Eliminar un par clave:valor 
del mi_perro['peso'] 
print(mi_perro) 

# Obtener las claves 
print(mi_perro.keys()) 

# Obtener los valores 
print(mi_perro.values())
```
