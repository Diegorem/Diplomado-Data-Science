## Leer tablas desde un PDF con Tabula

```python
#!pip install tabula-py[jpype]

from tabula import read_pdf
import pandas as pd

tablas = read_pdf(
    "4.1 sb2.pdf",
    pages="all",
    multiple_tables=False,
)

print(type(tablas))
```

### Parámetros
- pages='all': lee todas las páginas.
- multiple_tables=False: interpreta todo como una sola tabla.

```python
df = tablas[0]
df.head(10)
```

## Limpieza de datos extraídos

```python
df['Fecha'] = df['Fecha'].str.replace('\r', ' ') 
df['Ganador'] = df['Ganador'].str.replace('\r', ' ').str.rstrip('(0123456789)') 
df['Marcador'] = df['Marcador'].str.replace(r'\[.*\]', '', regex=True) 
df['Perdedor'] = df['Perdedor'].str.replace('\r', ' ').str.rstrip('(0123456789)') 
df['Estadio'] = df['Estadio'].str.replace('\r', ' ').str.rstrip('(0123456789)') 
df['Ciudad'] = df['Ciudad'].str.replace('\r', ' ').str.rstrip('(0123456789)')
```

Los PDFs suelen contener saltos de línea, referencias y caracteres especiales que deben limpiarse.
