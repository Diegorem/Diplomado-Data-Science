# Consumiendo APIs con Python
Notebook didáctico.

```python
import requests, json, pandas as pd
```

## ¿Qué es una API?
Una API permite intercambiar información estructurada entre aplicaciones. En Data Science es la forma preferida de obtener datos cuando existe una API oficial.

```python
response=requests.get('https://randomuser.me/api/', timeout=10)
print(response.status_code)
```

```python
datos=response.json()
print(json.dumps(datos, indent=2)[:1200])
```

## Acceder a datos del JSON

```python
u=datos['results'][0]
print(u['name']['first'],u['name']['last'])
print(u['email'])
```

## Obtener varios usuarios

```python
r=requests.get('https://randomuser.me/api/?results=10', timeout=10)
usuarios=r.json()['results']
for x in usuarios:
    print(x['name']['first'])
```

## Crear DataFrame

```python
df=pd.DataFrame([{'Nombre':x['name']['first'],'Apellido':x['name']['last'],'Pais':x['location']['country'],'Edad':x['dob']['age'],'Email':x['email']} for x in usuarios])
df.head()
```

## Exportar CSV

```python
df.to_csv('usuarios_random.csv',index=False)
```

## Ejercicio
Solicita 20 usuarios y calcula la edad promedio.

```python
r=requests.get('https://randomuser.me/api/?results=20', timeout=10)
usuarios=r.json()['results']
edades=[x['dob']['age'] for x in usuarios]
print('Edades:', edades)
print('Edad promedio:', round(sum(edades)/len(edades), 2))
```
