BeautifulSoup

BeautifulSoup convierte el HTML en un objeto fácil de navegar

```python
!pip install beautifulsoup4
```

```python
import requests
from bs4 import BeautifulSoup

url = "https://en.wikipedia.org/wiki/Titans_(2018_TV_series)"

page = requests.get(url, headers=headers)

soup = BeautifulSoup(
    page.content,
    "html.parser"
)
```

Buscar elementos con find()

Devuelve el primer elemento que coincide.

```python
titulo = soup.find("title")

print(titulo.text)
```

Buscar múltiples elementos con find_all()

Devuelve una lista.

```python
links = soup.find_all("a")

print(len(links))
```

```python
for link in links[:10]:
    print(link.get("href"))
```

Buscar por clase

Muchas páginas identifican elementos mediante clases CSS.

```python
titulos = soup.find_all(
    "td",
    class_="summary"
)
print(titulos)
```

Buscar usando expresiones regulares

Aquí aparece uno de los conceptos nuevos.

```python
import re

ids = soup.find_all(
    "th",
    id=re.compile(r"^ep\d+")
)

print(ids[:5])
```

Extraer información

Una vez encontrados los elementos:

```python
id_list = []

for item in ids:
    id_list.append(item.text)

print(id_list)
```

Obtener los nombres de los episodios

La lección utiliza:

```python
episodes = soup.find_all(
    "td",
    class_="summary"
)

episode_list = []

for episode in episodes:
    episode_list.append(
        episode.text
    )

print(episode_list)
```

Unir ambas listas

Finalmente:

```python
for numero, nombre in zip(id_list, episode_list):
    print(numero, "-", nombre)
```

Convertir a un DataFrame

Aunque el PDF solo lo menciona al final, una práctica muy común es guardar los datos en un DataFrame.

```python
import pandas as pd

df = pd.DataFrame({
    "Episodio": id_list,
    "Nombre": episode_list
})

print(df.head())
```
