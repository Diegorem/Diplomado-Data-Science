# Pandas: concat(), merge() y join()

Este notebook está diseñado para ejecutar cada celda y observar cómo funcionan las principales técnicas para combinar información en Pandas.


```python
import pandas as pd
import numpy as np

print("Pandas version:", pd.__version__)
```

## 1. Concatenación vertical con concat()

```python
enero = pd.DataFrame({
    'producto':['Laptop','Mouse'],
    'ventas':[10,25]
})

febrero = pd.DataFrame({
    'producto':['Teclado','Monitor'],
    'ventas':[15,8]
})

resultado = pd.concat([enero, febrero])
resultado
```

### Corregir índices con ignore_index

```python
resultado = pd.concat(
    [enero, febrero],
    ignore_index=True
)
resultado
```

## 2. Concatenación horizontal

```python
clientes = pd.DataFrame({
    'nombre':['Juan','Ana']
})

compras = pd.DataFrame({
    'total':[500,300]
})

pd.concat([clientes, compras], axis=1)
```

## 3. Outer Join usando concat

```python
df1 = pd.DataFrame({'A':[1,2]}, index=[0,1])
df2 = pd.DataFrame({'B':[10,20]}, index=[1,2])

pd.concat([df1, df2], axis=1)
```

## 4. Inner Join usando concat

```python
pd.concat(
    [df1, df2],
    axis=1,
    join='inner'
)
```

## 5. Left y Right Join con reindex

```python
left_join = pd.concat([df1, df2], axis=1).reindex(df1.index)
right_join = pd.concat([df1, df2], axis=1).reindex(df2.index)

print("LEFT JOIN")
display(left_join)

print("RIGHT JOIN")
display(right_join)
```

## 6. merge() básico

```python
clientes = pd.DataFrame({
    'id':[1,2,3],
    'nombre':['Juan','Ana','Pedro']
})

ventas = pd.DataFrame({
    'id':[1,2,3],
    'monto':[500,300,700]
})

pd.merge(clientes, ventas, on='id')
```

## 7. merge() con múltiples llaves

```python
ventas = pd.DataFrame({
    'cliente':[1,1,2],
    'mes':['Enero','Febrero','Enero'],
    'venta':[100,200,300]
})

pagos = pd.DataFrame({
    'cliente':[1,1,2],
    'mes':['Enero','Febrero','Enero'],
    'pagado':[True,False,True]
})

pd.merge(
    ventas,
    pagos,
    on=['cliente','mes']
)
```

## 8. Diferentes valores de how

```python
left = pd.DataFrame({
    'id':[1,2,3],
    'nombre':['Juan','Ana','Pedro']
})

right = pd.DataFrame({
    'id':[2,3,4],
    'ciudad':['CDMX','Monterrey','Guadalajara']
})

print("INNER")
display(pd.merge(left,right,on='id',how='inner'))

print("LEFT")
display(pd.merge(left,right,on='id',how='left'))

print("RIGHT")
display(pd.merge(left,right,on='id',how='right'))

print("OUTER")
display(pd.merge(left,right,on='id',how='outer'))
```

## 9. CROSS JOIN (producto cartesiano)

```python
clientes = pd.DataFrame({
    'cliente':['Juan','Ana']
})

productos = pd.DataFrame({
    'producto':['Laptop','Mouse']
})

pd.merge(
    clientes,
    productos,
    how='cross'
)
```

## 10. join() usando índices

```python
clientes = pd.DataFrame(
    {'nombre':['Juan','Ana']},
    index=[1,2]
)

ventas = pd.DataFrame(
    {'monto':[500,300]},
    index=[1,2]
)

clientes.join(ventas)
```

## 11. join() usando una columna

```python
df = pd.DataFrame({
    'id':[1,2,3],
    'nombre':['Juan','Ana','Pedro']
})

other = pd.DataFrame({
    'id':[1,2,3],
    'monto':[500,300,700]
})

df.join(
    other.set_index('id'),
    on='id'
)
```
