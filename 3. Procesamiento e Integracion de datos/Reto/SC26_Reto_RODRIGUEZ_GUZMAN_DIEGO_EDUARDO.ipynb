```python
import pandas as pd
import pandas as pd
```

```python
df1 = pd.read_csv("data1.csv")
print(df1.head())
df2 = pd.read_csv("data2.csv")
print(df2.head())
df3 = pd.read_csv("data3.csv")
print(df3.head())
df4 = pd.read_csv("data4.csv")
print(df4.head())
```

```python
datosUsuarios = pd.concat([df1, df2], ignore_index=True)

datosSO = pd.concat([df3, df4], ignore_index=True)

todo = pd.merge(datosUsuarios, datosSO, on='Employee_ID', how='outer')

print(todo.head())
```

```python
todo[' Computer_OS'] = todo[' Computer_OS'].fillna("No registrado")
todo[' Mobile_OS'] = todo[' Mobile_OS'].fillna("No registrado")
todo[' Computer_OS_version'] = todo[' Computer_OS_version'].fillna("No registrado")
todo[' Mobile_OS_version'] = todo[' Mobile_OS_version'].fillna("No registrado")
print(todo.head(30))
```

Consulta 1. Por cada lugar ("Hometown"), cuántos hombres y cuántas mujeres hay.

```python
lugarFyM = pd.pivot_table(
    todo,
    index="Hometown",
    columns="Gender",
    aggfunc="size",
)
lugarFyM.head()
```

Consulta 2: El número de empleados que utilizan cada Sistema Operativo de Escritorio

```python
SOescritorio = pd.pivot_table(
    todo,
    index=" Computer_OS",
    values="Employee_ID",
    aggfunc="count"
)

SOescritorio.head()
```

Consulta 3: El promedio de la edad y de nivel de educación de las personas por cada Sistema Operativo en su dispositivo móvil.

```python
EdadNivEdProm = todo.groupby([' Mobile_OS'])[['Age','Education_Level']].mean()
EdadNivEdProm.head()
```

Consulta 4: El número total de tickets generados por las personas que usan cada uno de los sistemas operativos de escritorio. Como columnas, en un nivel superior agrupa por género, y en un nivel inferior por estado civil.

```python
tickOS = pd.pivot_table(
    todo,
    index=' Computer_OS',
    columns=['Gender', 'Relationship_Status'],
    values='Employee_ID',
    aggfunc='count',
)

tickOS.head()
```

Consulta 5: Genera un DataFrame que contenga como filas, cada uno de los lugares en la columna "Hometown", y como columnas de primer nivel el estado relacional, y como segundo nivel el género. Los valores a mostrar en cada intersección de fila con columna deben ser el promedio de la columna "Attrition_rate" que cumplan con esos valores de fila y columna.

```python
df5 = pd.pivot_table(
    todo,
    index='Hometown',
    columns=['Relationship_Status','Gender'],
    values="Attrition_rate",
    aggfunc="mean"
)
df5.head()
```

Enviar el resultado de cada solicitud de información a un mismo archivo XLS[X], donde el resultado de cada consulta debe ir en su propia hoja de trabajo. Cada hoja de trabajo debe llevar por nombre c1, c2, c3, c4 y c5, cada una correspondiente a cada una de las 5 consultas planteadas.

```python
with pd.ExcelWriter('consultas.xlsx') as writer:
    lugarFyM.to_excel(writer, sheet_name='c1')
    SOescritorio.to_excel(writer, sheet_name='c2')
    EdadNivEdProm.to_excel(writer, sheet_name='c3')
    tickOS.to_excel(writer, sheet_name='c4')
    df5.to_excel(writer, sheet_name='c5')
```
