```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

# 1. Preparación de Datos (Simulación)

```python
years = [str(y) for y in range(2013, 2024)]
pop_values = [
    118343573.0,
    119784261.0,
    121072306.0,
    122251351.0,
    123400057.0,
    124573711.0,
    125762982.0,
    126799054.0,
    127648148.0,
    128613117.0,
    129739759.0,
]
mexico = pd.DataFrame({'Population': pop_values}, index=years)

gender_mex = pd.DataFrame(
    {
        'Year': [2019, 2019, 2020, 2020, 2021, 2021, 2022, 2022, 2023, 2023],
        'Sex': [
            'female',
            'male',
            'female',
            'male',
            'female',
            'male',
            'female',
            'male',
            'female',
            'male',
        ],
        'Population': [
            65172136,
            62403393,
            64540634,
            61473390,
            64912365,
            61792773,
            65345824,
            62158301,
            67075455,
            62664304,
        ],
    }
)
```

# 2. Configuración Global de Estilo

```python
plt.rcParams['figure.figsize'] = (10, 5)
sns.set_theme(palette='pastel', style='whitegrid')
```

# 3. Ejemplo Matplotlib Pyplot

```python
plt.figure()
plt.plot(
    mexico.index,
    mexico['Population'],
    color='r',
    linestyle='--',
    marker='o',
    label='Población',
)
plt.title('Población de México (Matplotlib)')
plt.xlabel('Año')
plt.ylabel('Población')
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

# 4. Ejemplo Seaborn con Semántica (Hue/Style)

```python
plt.figure()
sns.lineplot(
    data=gender_mex,
    x='Year',
    y='Population',
    hue='Sex',
    style='Sex',
    marker='o',
)
plt.title('Población por Sexo en México (Seaborn)')
plt.tight_layout()
plt.show()
```

# 5. Ejemplo Interfaz Orientada a Objetos (Subplots)

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

mexico.plot(ax=ax1, legend=False, color='green', marker='s')
ax1.set_title('Pandas en Subplot 1')
ax1.tick_params(axis='x', rotation=45)

sns.lineplot(
    data=gender_mex, x='Year', y='Population', hue='Sex', ax=ax2, marker='d'
)
ax2.set_title('Seaborn en Subplot 2')

plt.tight_layout()
plt.show()
```
